# Preface

These notebooks grew out of a series of four lectures on *Computational Aspects of Complex Reflection Groups*.  Its subject is a single algorithm, and how far it can be taken.

The **orbit algorithm**, breadth-first search in the graph of a group action, finds the orbit of a point under a group given by a list of generators.  With small extensions, it also finds words for the group elements, a transversal of the stabilizer, Schreier generators of the stabilizer, the edges of the action graph, and the permutations induced by the action.  And with a suitable choice of action, many things that do not look like orbits turn out to be orbits: the elements, the conjugacy classes and the subgroups of a group, the partitions of a number, the shapes of the parabolic subgroups of a Coxeter group, its involutions, the cosets of a subgroup of a group given by a presentation, and a basis of a module.  You'll be amazed by what all qualifies as an orbit.

## Overview

* **Part 1, *Algorithms: A Glimpse of CGT***, introduces tree traversal by breadth-first and depth-first search [CLRS22], and develops the orbit algorithm and its variations for permutation groups: elements, words, conjugacy classes and subgroups, stabilizers and Schreier generators, and stabilizer chains, which give the order of a group, random elements and a membership test [Sim70].  The examples include the symmetric groups, the Mathieu group $M_{24}$ [CCNPW85] and the Rubik's cube group [Joy08].
* **Part 2, *Orbits: Finite Coxeter Groups in Action***, constructs a finite Coxeter group as a permutation group on its root system [Hum90], and computes lengths, reduced words, longest elements, parabolic subgroups and their distinguished coset representatives, the shapes of the parabolic subgroups and the complements in their normalizers [How80], and the conjugacy classes, by way of their elements of minimal length [GP93], [GP00].
* **Part 3, *Cosets: Enumerating Complex Reflection Groups***, turns the presentation of a complex reflection group [ST54], [BMR98], [LT09] into a permutation group by **coset enumeration** [TC36], [Neu82], [Sim94].  The coincidences that arise along the way are managed by **Union-Find** [GF64], [Tar75].
* **Part 4, *Vectors: Enumerating Modules and Hecke Algebras***, replaces points by vectors.  The linear version of the orbit algorithm, **spinning**, constructs Specht modules [Jam78], [Sag01] and the representations induced from a stabilizer; and a linear version of Union-Find prepares the ground for **vector enumeration** [Lin91] in Iwahori-Hecke algebras [GP00].

## Prerequisites

The reader should be familiar with the basic notions of group theory (group actions, cosets, conjugacy) and of linear algebra.  Parts 2 and 4 assume some familiarity with Coxeter groups and their root systems, as in [Hum90] or [GP00, Chapter 1].  No prior knowledge of Julia [BEKS17] is needed: the language features are introduced as they are used, in boxes marked with the Julia logo.

## Scope

This is not a comprehensive account of Computational Group Theory.  It follows one idea through a range of examples, and leaves out much of the standard material, or touches on it only in the exercises: the Schreier-Sims algorithm with bases and strong generating sets [Sim70], [Ser03], backtrack searches, presentations of subgroups, and the MeatAxe [Par84].  For systematic treatments, see the books [But91], [Sim94], [Ser03] and [HEO05].  Production quality implementations of these algorithms, and of many more, are available in GAP [GAP], and, for Coxeter groups, reflection groups and Hecke algebras, in CHEVIE [GHLMP96], [Mic15].

## Current Directions

The algorithms in these notebooks are classical, but the field is very much alive.

* **Software.**  The Julia ecosystem now has a full computer algebra system, OSCAR [DEFHJ25], which includes GAP, and CHEVIE has been rewritten in Julia as Chevie.jl [Chevie.jl].
* **Permutation groups.**  Backtrack search is still the method of choice for centralizers, normalizers and intersections, and partition backtrack [Leo91] is still being refined [JPW19].  Babai's quasipolynomial algorithm for graph isomorphism [Bab16] builds on the group theoretic approach of Luks [Luk82], with orbits and stabilizers at its core.
* **Matrix groups.**  For groups of matrices over finite fields, stabilizer chains are often too long.  The matrix group recognition project combines the MeatAxe with the structure theory of finite groups [OBr11].
* **Presentations.**  Coset enumeration has been extended from groups to monoids and semigroups, for example in libsemigroups [libsemigroups] (cf. Exercise 3.1).
* **Reflection groups and Hecke algebras.**  Complex reflection groups and their braid groups [Bro10] and Hecke algebras remain an active field, where computations, such as the enumeration of vectors in a Hecke algebra (Part 4), have helped to prove the BMR freeness conjecture, which now holds for all exceptional groups; see [MP17] for the 2-reflection groups.  For the combinatorics of Coxeter groups, see also [BB05].

## How to Use These Notebooks

Each chapter is a Jupyter notebook, which can be run in Julia, locally or online through Binder, from the repository https://github.com/gpfeiffer/orbits-julia.  The code is meant to be read, run and modified: the notebooks develop the algorithms themselves, and later chapters load the algorithms of the earlier ones from the package [OrbitAl](https://github.com/gpfeiffer/OrbitAl.jl).

The exercises at the end of each chapter are numbered.  Where an exercise asks for a function, the notebook provides a stub to complete, and a check that compares the result with the implementation in OrbitAl.  The solutions are collapsed in the notebooks, and collected in an appendix in the PDF version.
